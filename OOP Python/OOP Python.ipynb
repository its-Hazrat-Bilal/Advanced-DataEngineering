### Explination of self in python
 
Self is a reference to the current instance of the class. It is used to access variables that belong to the class. It binds the attributes with the given arguments. The self parameter is a reference to the current instance of the class and is used to access variables that belong to the class. It is explicitly included as the first parameter in method definitions, allowing methods to access and modify the instance's attributes and other methods.

```python
# Self Detailed Example

class Car:
    # This method sets up the car's identity
    def setup(self, car_color, car_brand):
        # 'self' means "this specific car object".
        # We are attaching a color and brand to THIS specific car.
        self.color = car_color
        self.brand = car_brand

    # This method prints the car's details
    def show_details(self):
        # 'self.brand' tells Python: "Look inside MYSELF and find MY brand."
        # 'self.color' tells Python: "Look inside MYSELF and find MY color."
        print("I am a " + self.color + " " + self.brand)

# STEP 1: Create two completely separate cars

car1 = Car()
car2 = Car()

# STEP 2: Give each car its own unique data

# When we call car1.setup(), Python secretly passes 'car1' into 'self'.
# It is exactly like writing: car1.color = "Red" and car1.brand = "Ferrari"
car1.setup("Red", "Ferrari")

# When we call car2.setup(), Python secretly passes 'car2' into 'self'.
# It is exactly like writing: car2.color = "Blue" and car2.brand = "Tesla"
car2.setup("Blue", "Tesla")

# STEP 3: Make each car speak for itself


# 1. We call show_details on car1
car1.show_details()  
# Inside the class, 'self' becomes 'car1'.
# It prints: "I am a Red Ferrari"

# 2. We call show_details on car2
car2.show_details()  
# Inside the class, 'self' becomes 'car2'.
# It prints: "I am a Blue Tesla"
```

### More Arguments

```python
import math

class Point:
    def move (self, x, y):
        self.x = x
        self.y = y
    
    def reset(self):
        self.move(0, 0)  # calling the move method to reset the point to (0, 0)
    
    def calculate_distance(self, other_point):
        return math.sqrt(
            (self.x - other_point.x) ** 2 +
            (self.y - other_point.y) ** 2
        )
    
# How to use it???
point1 = Point()
point2 = Point()

point1.reset()  # point1 is now at (0, 0)
point2.move(3, 4)  # point2 is now at (3, 4)

print(point1.calculate_distance(point2))  # prints 5.0, which is the distance between (0, 0) and (3, 4)

assert(point2.calculate_distance(point1) == point1.calculate_distance(point2))  # distance is symmetric

point1.move(3, 4) # point1 is now at (3, 4)
print(point1.calculate_distance(point2))  # prints 0.0, since both points are at (3, 4)
print(point1.calculate_distance(point1))  # prints 0.0, since the distance to itself is zero
```

Question 2: The Bank Account

Create a class called BankAccount.
Write a method to deposit money into the account.Write a method to withdraw money from the account.
Write a method to view the current balance.

Note: Assume the account starts with $0 before the first deposit.

```python
class BankAccount:
    def deposit(self, amount):
        self.amount = amount

        return self.amount
    
    def withdraw(self, amount):
        if amount > self.amount:
            return "Insufficient funds"
        else:
            self.amount -= amount

            return self.amount
    
    def check_balance(self):
        return self.amount
    
my_account = BankAccount()

my_account.deposit(1000)
print("current balance is " + str(my_account.check_balance()))

my_account.withdraw(500)
print("current balance is " + str(my_account.check_balance()))
```

Question 3: The Score Tracker

Create a class called GameScore.

Write a method to log a scoring play (adds a specific number of points to the score).

Write a method to penalize the team (subtracts a specific number of points from the score).

Write a method to reset the score back to 0.Write a method to print out the final score.

```python
class GameScore:
    def scoring_log(self, point):
        self.point = point

    def penalize(self, penalty):
        self.penalty = penalty
        self.total_score = self.point - self.penalty
    
    def reset(self):
        self.point = 0
        self.penalty = 0
        self.total_score = 0

game = GameScore()
game.scoring_log(10)
game.penalize(2)
game.reset()
```

Most object-oriented programming languages have the concept of a constructor,  a special method that creates and initializes the object when it is created.

like in c++

    private:
    int x;

    Car(int num) {

    }

>  Python is a little different; it has a constructor and an initializer. The constructor function is rarely used unless you're doing something exotic.

The Python initialization method is the same as any other method, except it has a special name, __init__. The leading and trailing double underscores mean this is a special method that the Python interpreter will treat as a special case.

```python
# initialization function

class Point:
    def __init__ (self, x, y):
        self.move(x, y)
    
    def move(self, x, y):
        self.x = x
        self.y = y
    
    def reset(self):
        self.move(0, 0)

# constructing a point
point = Point(3, 4)  # point is now at (3, 4)
print(point.x, point.y)  # prints 3 4
```

```python
# Constructor is a special method that is automatically called when an object is created. It is used to initialize the object's attributes. In Python, the constructor method is defined using the `__init__` method.

class Car:
    # Default constructor
    def __init__(self):
        print("Constructor called")

c = Car()
```

### Class & Instance Arrtibutes

Class variables are shared across all instances of a class. They are defined within the class construction but outside of any instance methods. Class variables are used to store data that is common to all instances of a class.


Instance Arrtibues/ Variables are unique to each instance of a class. They are defined within the constructor method and are prefixed with `self.` to indicate that they belong to the specific instance. 

```python
class Student:
    school_name = "ABC School"  # class variable
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("John", 85)
s2 = Student("Alice", 90)
print(s1.name + " scored " + str(s1.marks) + " in " + s1.school_name)
print(s2.name + " scored " + str(s2.marks) + " in " + s2.school_name)
```

```python
# When a paramter is already passed...

class Student:
    school_name = "ABC School"
    
    # Set "Ali" as the default value for the name parameter
    # Note: I switched the order so marks is required, but name is optional
    def __init__(self, marks, name="Ali"): 
        self.name = name
        self.marks = marks

# 1. Uses the default "Ali"
s1 = Student(85) 

# 2. Overwrites the default with "Alice"
s2 = Student(90, "Alice") 

print(f"{s1.name} scored {s1.marks} in {s1.school_name}") # Ali scored 85 in ABC School
print(f"{s2.name} scored {s2.marks} in {s2.school_name}") # Alice scored 90
```
