### Classes and Objects in python (same as in cpp)

```python
class myfirstclass:
    pass

# simplest class
```

```python
# objects in python are instances of classes. A class is a blueprint for creating objects.
# An object has properties and methods(functions) associated with it.
# Almost everything in Python is an object, with its properties and methods.

a = myfirstclass()  # creating an object of the class
b = myfirstclass()  # creating another object of the class
```

```python
print(a)  # printing the object
print(b)  # printing the object
```

```python
# adding attributes to the class

class point:
    pass
```

```python
p1 = point()  # creating an object of the class
p2 = point()  # creating another object of the class

p1.x = 5  # adding an attribute to the object
p1.y = 10  # adding another attribute to the object

p2.x = 15  # adding an attribute to the object
p2.y = 20  # adding another attribute to the object

print(p1.x, p1.y)  # printing the attributes of the object
print(p2.x, p2.y)  # printing the attributes of the object

# Here in python, we can add attributes to the object of the class.
# We can also add methods to the class.
# but in cpp you cant do it like this
# example
# class point() {
#     public:
#       int x, y;
# }
# int main() {
#     point p1;  // creating an object of the class
#     p1.x = 5;  // adding an attribute to the object
#     p1.y = 10;  // adding another attribute to the object
#     point p2;  // creating another object of the class
#     p2.x = 15;  // adding an attribute to the object
#     p2.y = 20;  // adding another attribute to the object
#     return 0;
# }
```

This code creates an empty Point class with no data or behaviors. Then it creates
two instances of that class and assigns each of those instances x and y coordinates to
identify a point in two dimensions. All we need to do to assign a value to an attribute
on an object is use the syntax < object > .< attribute > = < value >. This is sometimes
referred to as dot notation. The value can be anything: a Python primitive, a
built-in data type, another object. It can even be a function or another class!

### Making a class do something

```python
class point:
    def reset(self):
        self.x = 0
        self.y = 0

p = point()  # creating an object of the class
p.reset()  # calling the method of the class

print(p.x, p.y)  # printing the attributes of the object
```

### Explination of self in python
 
Self is a reference to the current instance of the class. It is used to access variables that belong to the class. It binds the attributes with the given arguments. The self parameter is a reference to the current instance of the class and is used to access variables that belong to the class. It is explicitly included as the first parameter in method definitions, allowing methods to access and modify the instance's attributes and other methods.

```python
# Self Detailed Example

class Car:
    # This method sets up the car's identity
    def setup(self, car_color, car_brand):
        # 'self' means "this specific car object".
        # We are attaching a color and brand to THIS specific car.
        self.color = car_color
        self.brand = car_brand

    # This method prints the car's details
    def show_details(self):
        # 'self.brand' tells Python: "Look inside MYSELF and find MY brand."
        # 'self.color' tells Python: "Look inside MYSELF and find MY color."
        print("I am a " + self.color + " " + self.brand)

# STEP 1: Create two completely separate cars

car1 = Car()
car2 = Car()

# STEP 2: Give each car its own unique data

# When we call car1.setup(), Python secretly passes 'car1' into 'self'.
# It is exactly like writing: car1.color = "Red" and car1.brand = "Ferrari"
car1.setup("Red", "Ferrari")

# When we call car2.setup(), Python secretly passes 'car2' into 'self'.
# It is exactly like writing: car2.color = "Blue" and car2.brand = "Tesla"
car2.setup("Blue", "Tesla")

# STEP 3: Make each car speak for itself


# 1. We call show_details on car1
car1.show_details()  
# Inside the class, 'self' becomes 'car1'.
# It prints: "I am a Red Ferrari"

# 2. We call show_details on car2
car2.show_details()  
# Inside the class, 'self' becomes 'car2'.
# It prints: "I am a Blue Tesla"
```
