### Explination of self in python
 
Self is a reference to the current instance of the class. It is used to access variables that belong to the class. It binds the attributes with the given arguments. The self parameter is a reference to the current instance of the class and is used to access variables that belong to the class. It is explicitly included as the first parameter in method definitions, allowing methods to access and modify the instance's attributes and other methods.

```python
# Self Detailed Example

class Car:
    # This method sets up the car's identity
    def setup(self, car_color, car_brand):
        # 'self' means "this specific car object".
        # We are attaching a color and brand to THIS specific car.
        self.color = car_color
        self.brand = car_brand

    # This method prints the car's details
    def show_details(self):
        # 'self.brand' tells Python: "Look inside MYSELF and find MY brand."
        # 'self.color' tells Python: "Look inside MYSELF and find MY color."
        print("I am a " + self.color + " " + self.brand)

# STEP 1: Create two completely separate cars

car1 = Car()
car2 = Car()

# STEP 2: Give each car its own unique data

# When we call car1.setup(), Python secretly passes 'car1' into 'self'.
# It is exactly like writing: car1.color = "Red" and car1.brand = "Ferrari"
car1.setup("Red", "Ferrari")

# When we call car2.setup(), Python secretly passes 'car2' into 'self'.
# It is exactly like writing: car2.color = "Blue" and car2.brand = "Tesla"
car2.setup("Blue", "Tesla")

# STEP 3: Make each car speak for itself


# 1. We call show_details on car1
car1.show_details()  
# Inside the class, 'self' becomes 'car1'.
# It prints: "I am a Red Ferrari"

# 2. We call show_details on car2
car2.show_details()  
# Inside the class, 'self' becomes 'car2'.
# It prints: "I am a Blue Tesla"
```

### More Arguments

```python
import math

class Point:
    def move (self, x, y):
        self.x = x
        self.y = y
    
    def reset(self):
        self.move(0, 0)  # calling the move method to reset the point to (0, 0)
    
    def calculate_distance(self, other_point):
        return math.sqrt(
            (self.x - other_point.x) ** 2 +
            (self.y - other_point.y) ** 2
        )
    
# How to use it???
point1 = Point()
point2 = Point()

point1.reset()  # point1 is now at (0, 0)
point2.move(3, 4)  # point2 is now at (3, 4)

print(point1.calculate_distance(point2))  # prints 5.0, which is the distance between (0, 0) and (3, 4)

assert(point2.calculate_distance(point1) == point1.calculate_distance(point2))  # distance is symmetric

point1.move(3, 4) # point1 is now at (3, 4)
print(point1.calculate_distance(point2))  # prints 0.0, since both points are at (3, 4)
print(point1.calculate_distance(point1))  # prints 0.0, since the distance to itself is zero
```

Create a class called BankAccount.
Write a method to deposit money into the account.Write a method to withdraw money from the account.
Write a method to view the current balance.

Note: Assume the account starts with $0 before the first deposit.

```python
class BankAccount:
    def deposit(self, amount):
        self.amount = amount

        return self.amount
    
    def withdraw(self, amount):
        if amount > self.amount:
            return "Insufficient funds"
        else:
            self.amount -= amount

            return self.amount
    
    def check_balance(self):
        return self.amount
    
my_account = BankAccount()

my_account.deposit(1000)
print("current balance is " + str(my_account.check_balance()))

my_account.withdraw(500)
print("current balance is " + str(my_account.check_balance()))
```

### Static Method

Methods that don't use the self parameter (work at class level, not instance level) are called static methods. They are defined using the @staticmethod decorator.

```python
class Std:
    @staticmethod # Decorator to define a static method
    def college_name():
        return "XYZ College"

s = Std()
print(s.college_name())

# See no error even the self parameter is not passed to the static method. This is because static methods do not require an instance of the class to be called. They can be called directly on the class itself, without creating an object.
```

### class method
It is a bound to the class & receives the class as an implicit first argument.

Note: static method can't access or modify class state & generally for utility

```python
class Dog:
    species = "Canine"  # Class variable (Shared by all dogs)

    def __init__(self, name):
        self.name = name  # Instance variable (Unique to each dog)

    # 1. Instance Method (Works with the specific dog's name)
    def bark(self):
        return f"{self.name} says woof!"

    # 2. Class Method (Works with the whole species)
    @classmethod
    def change_species(cls, new_species):
        cls.species = new_species
        return f"Species changed to {cls.species}"

    # 3. Static Method (General helper function, needs no dog data)
    @staticmethod
    def run_speed(distance, time):
        return f"Speed is {distance / time} m/s"


# --- How to use them ---

dog1 = Dog("Buddy")
dog2 = Dog("Max")

# 1. Instance Method: Uses 'self' to know WHICH dog is barking
print(dog1.bark())  # Output: Buddy says woof!
print(dog2.bark())  # Output: Max says woof!

# 2. Class Method: Uses 'cls' to update the variable for ALL dogs
print(Dog.change_species("Wolf"))  # Output: Species changed to Wolf

# 3. Static Method: Just a regular math tool inside the class
print(Dog.run_speed(100, 10))  # Output: Speed is 10.0 m/s
```

### Property

```python
# Property method is used to define a method as a property of the class. It allows you to access the method like an attribute, without needing to call it with parentheses.
# This is useful for creating read-only attributes or computed properties.

class Student:
    # list of marks?
    def __init__(self, name, a, b, c):
        self.name = name
        self.a = a
        self.b = b
        self.c = c

    @property
    def percentage(self):
        return f"{str((self.a + self.b + self.c) / 3)} %"
    
s = Student("John", 85, 90, 78)
print(s.percentage)
```

## Polymorphism
Polymorphism is a concept in object-oriented programming that allows objects of different classes to be treated as objects of a common superclass. It enables a single interface to represent different underlying forms (data types). In Python, polymorphism can be achieved through method overriding and operator overloading.


## Method Overriding
Method overriding occurs when a subclass provides a specific implementation of a method that is already defined in its superclass. The overridden method in the subclass has the same name, return type, and parameters as the method in the superclass. This allows the subclass to provide its own behavior while still maintaining the interface defined by the superclass.

### Operator Overloading
Operator overloading allows you to define how operators (like +, -, *, etc.) behave for objects of your custom classes. By implementing special methods (also known as "dunder" methods) in your class, you can specify the behavior of these operators when applied to instances of your class.


```python
# Operator Overlaoding

class Comples:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def shownumber(self):
        return f"{self.real} + {self.imag}i"

    def __add__(self, num2):
        new_real = self.real + num2.real
        new_imag = self.imag + num2.imag
        return Comples(new_real, new_imag)

    def __sub__(self, num2):
        new_real = self.real - num2.real
        new_imag = self.imag - num2.imag
        return Comples(new_real, new_imag)

n = Comples(3, 4)
print(n.shownumber())  # Output: 3 + 4i

n2 = Comples(1, 2)
print(n2.shownumber())  # Output: 1 + 2i

n3 = n + n2  # This will call the __add__ method
print(n3.shownumber())  # Output: 4 + 6i

n4 = n - n2  # This will call the __sub__ method
print(n4.shownumber())  # Output: 2 + 2i
```
