```python
# The goal of this exercise is to convert a string to a new string where each character in the new string is "(" if that character appears only once in the original string, or ")" if that character appears more than once in the original string. Ignore capitalization when determining if a character is a duplicate.

# Examples
# "din"      =>  "((("
# "recede"   =>  "()()()"
# "Success"  =>  ")())())"
# "(( @"     =>  "))((" 
# Notes
# Assertion messages may be unclear about what they display in some languages. If you read "...It Should encode XXX", the "XXX" is the expected result, not the input!

def duplicate_encode(word):
    result = ""
    for i in word:
        if word.lower().count(i.lower()) > 1:
            i = ")"
            result += i
        else:
            i = "("
            result += i
    return result

duplicate_encode("(( @")
```

```python
# #leet code
# You are given a positive integer n. Determine whether n is divisible by the sum of the following two values:
# The digit sum of n (the sum of its digits).
# The digit product of n (the product of its digits).
# Return true if n is divisible by this sum; otherwise, return false.
# Example 1:
# Input: n = 99
# Output: true
# Explanation:
# Since 99 is divisible by the sum (9 + 9 = 18) plus product (9 * 9 = 81) of its digits (total 99), the output is true.
# Example 2:
# Input: n = 23
# Output: false
# Explanation:
# Since 23 is not divisible by the sum (2 + 3 = 5) plus product (2 * 3 = 6) of its digits (total 11), the output is false.

# Constraints:
# 1 <= n <= 106

class Solution(object):
    def checkDivisibility(self, n):
        """
        :type n: int
        :rtype: bool
        """
        sum_digits, product_digits = 0, 1
        n = str(n)
        for i in n:
            sum_digits += int(i)
            product_digits *= int(i)
        if (sum_digits + product_digits % int(n) == 0):
            return True
        else:
            return False
    

s = Solution()
check = s.checkDivisibility(9)
print(check)
```

```python
# same as previous but more optimized

class Solution(object):
    def checkDivisibility(self, n):
        original_n = n
        sum_digits, product_digits = 0, 1

        while n > 0:
            digit = n % 10
            sum_digits += digit
            product_digits *= digit
            n //= 10
        return original_n % (sum_digits + product_digits) == 0


s = Solution()
check = s.checkDivisibility(15)
print(check)
```

```python
class Solution(object):
    def twoSum(self, nums, target):
        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] + nums[j] == target:
                    return [i, j]

s = Solution()
s.twoSum([2, 7, 11, 15], 17)
```

```python
# Write a Python decorator `@timer` that prints how long a function takes to run.



from time import time


def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Function {func.__name__} took {end - start:.2f} seconds to run.")
        return result
    return wrapper

timer([1, 2, 3, 4, 5])
```

```python
# Palandrome Problem Leet code

def check (no):
    if no == 0:
        return True 
    main_no = no
    reverse_no = 0
    while no > 0:
        digit = no % 10
        reverse_no = reverse_no * 10 + digit
        no //= 10

    return reverse_no == main_no



print(check(-121))
```

```python
# Roman to Integer

class Solution(object):
    def romanToInt(self, s):
        """
        :type s: str
        :rtype: int
        """
        dect = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
        total = 0
        for i in range(len(s)):
            if i + 1 < len(s) and dect[s[i]] < dect[s[i + 1]]:
                total -= dect[s[i]]
            else:
                total += dect[s[i]]
        return total
s = Solution()
print(s.romanToInt("MCMXCIV"))
```

```python
def is_full(top: int, size: int) -> bool:
    return top == size - 1


def is_empty(top: int) -> bool:
    return top == -1


def push(a: list, top: int, size: int) -> int:
    if is_full(top, size):
        print("Stack Overflow! Cannot add more elements.")
    else:
        no = int(input("Enter an element: "))
        top += 1
        a[top] = no
        print(f"{no} pushed onto stack.")
    return top


def pop(a: list, top: int) -> int:
    if is_empty(top):
        print("Stack Underflow! Nothing to pop.")
    else:
        print(f"{a[top]} popped from stack.")
        top -= 1
    return top


def peek(a: list, top: int) -> None:
    if is_empty(top):
        print("Stack is empty.")
    else:
        print(f"Top element is: {a[top]}")


def traverse(a: list, top: int) -> None:
    if is_empty(top):
        print("Stack is empty.")
        return
    print("Stack elements: ", end="")
    for i in range(top + 1):
        print(a[i], end="\t")
    print()


def main():
    size = 10
    a = [0] * size  # Fixed-size array initialized with 0s
    top = -1

    while True:
        print("\n--- STACK MENU ---")
        print("1. Push element")
        print("2. Pop element")
        print("3. Peek top element")
        print("4. Traverse stack")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            top = push(a, top, size)
        elif choice == "2":
            top = pop(a, top)
        elif choice == "3":
            peek(a, top)
        elif choice == "4":
            traverse(a, top)
        elif choice == "5":
            print("Exiting program...")
            break
        else:
            print("Invalid choice! Please try again.")
```

```python
# Complete the solution so that it strips all text that follows any of a set of comment markers passed in.
# Any whitespace at the end of the line should also be stripped out.
# Example:
# Given an input string of:

# apples, pears # and bananas
# grapes
# bananas !apples
# The output expected would be:

# apples, pears
# grapes
# bananas
def strip_comments(strng, markers):
    lines = strng.splitlines() 
    new_lines = []
    
    for i in lines:
        for marker in markers:
            if marker in i:
                i = i.split(marker)[0]
        
        new_lines.append(i.rstrip()) 
        
    return "\n".join(new_lines)

p = """apples, pears # and bananas
grapes
bananas !apples"""

print(strip_comments(p, ["#", "!"]))
```
